```python
# TEXT ANALYSIS

text = """Python is a powerful programming language.
Python is used for Artificial Intelligence and Machine Learning.
Learning Python helps students build useful projects."""

# Convert text to lowercase
text = text.lower()

# Remove punctuation
for symbol in ".,!?":
    text = text.replace(symbol, "")

# Split text into words
words = text.split()

# Count words
word_count = len(words)

# Count characters excluding spaces
character_count = len(text.replace(" ", "").replace("\n", ""))

# Count vowels
vowels = "aeiou"
vowel_count = 0

for char in text:
    if char in vowels:
        vowel_count += 1

# Count word frequency
word_frequency = {}

for word in words:
    if word in word_frequency:
        word_frequency[word] += 1
    else:
        word_frequency[word] = 1

# Find repeated words
repeated_words = {}

for word, count in word_frequency.items():
    if count > 1:
        repeated_words[word] = count

# Display results
print("TEXT ANALYSIS ")

print("\nOriginal Text:")
print(text)

print("\nTotal Words:", word_count)
print("Total Characters:", character_count)
print("Total Vowels:", vowel_count)

print("\nWord Frequency:")
for word, count in word_frequency.items():
    print(word, ":", count)

print("\nRepeated Words:")
if len(repeated_words) > 0:
    for word, count in repeated_words.items():
        print(word, ":", count)
else:
    print("No repeated words found.")
```
